# Assignment Problems: Day 15

The following problems require access to the file named ‘Dataset_Day15.csv’, provided with this assignment.

The data contains transaction details with 7501 rows. Each row is a grocery transaction from the market with the unique items list for each transaction.

```python
import pandas as pd
from csv import reader
from mlxtend.frequent_patterns import apriori
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import association_rules
```

* Read the data and remove all null or empty values.

```python
groceries =[]
with open('Dataset_Day15.csv','r') as csvfile:
    read = reader(csvfile)
    for row in read:
        row_new = [x for x in row if x!='NaN' and x!='']
        groceries.append(row_new)
groceries[0:5]
```

* Transform the data using TransactonEncoder to perform Market Basket Analysis.

```python
encoder = TransactionEncoder()
transactions = encoder.fit(groceries).transform(groceries)
itemsets = pd.DataFrame(transactions, columns = encoder.columns_)
itemsets
```

```python
itemsets.info()
```

* Use min_support = 0.02 to find frequent itemsets.

```python
frequent_itemsets = apriori(itemsets, min_support = 0.02, use_colnames = True)
frequent_itemsets['length'] = frequent_itemsets['itemsets'].str.len()
frequent_itemsets.sort_values('support', ascending = False)
```

```python
frequent_itemsets.groupby('length')['support'].describe()
```

* Use the frequent itemsets to create association rules (take min_threshold as 15%) and evaluate the rules for the following metrics
* * Find top 5 antecedent -> consequent rules based on ‘conviction’, ‘leverage’, ‘lift’. Explain your findings.
* * Find top 2 & bottom 2 antecedent -> consequent rules based on ‘zhang’s metric’. Explain your findings.

```python
rules = association_rules(frequent_itemsets, min_threshold=0.15, metric="confidence")
rules
```
