```python
# question - 1
# Write a Python program to check if a given number is prime or not.

def p_or_not(x):
    for i in range(2,x//2):
        if x%i == 0.0 :
            return False
    if x==1:
        return False
    else:
        return True


x=int(input('Please enter a number : '))
if p_or_not(x):
    print(f'{x} is a prime number.')
else:
    print(f'{x} is not a prime number.')
```

```python
# question - 2
# Write a function to input a number below 50 and print it’s factorial.

def fact(x):
    y=1
    for i in range(1,x+1):
        y=y*i
    return y


x=int(input('Please enter a number (<50) : '))
print(f'So factorial of {x} is {fact(x)}')
```

```python
# question - 3
# Find the count of occurrence of each city in the dataset and create a dictionary
# such that the keys are the unique city names, and the values are their count of occurrence.

import csv as d

with open('Dataset_Day1.csv') as f:
    r=d.reader(f)
    x={
        "city": [],
        "count" : []
    }
    for row in r :
        if row[0] in x["city"]:
            i=x["city"].index(row[0])
            x["count"][i] += 1
        else:
            x["city"].append(row[0])
            i=x["city"].index(row[0])
            x["count"].append(1)
x["count"][0]="count"
for n in x["city"]:
    i=x["city"].index(n)
    print(f' {x["city"][i]}  =  {x["count"][i]}')
```

```python
# question - 4
# Create a set of all the cities with at-least one Apollo hospital.

import csv as d

with open('Dataset_Day1.csv') as f:
    r=d.reader(f)
    x=[]
    name="apollo"
    for row in r :
        if row[1].lower().find(name)>=0:
            if row[0] not in x:
                x.append(row[0])
for n in x:
    print(n)
```
