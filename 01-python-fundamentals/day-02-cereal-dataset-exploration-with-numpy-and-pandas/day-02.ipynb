```python
import pandas as pd
import numpy as np
import csv as d
data = pd.read_csv("Dataset_Day2.csv")
data
```

```python
# question-1
# Find all unique values of columns: “name”, “mfr”, “vitamins” and store them in separate numpy arrays.

name = data["name"].unique()
mfr = data["mfr"].unique()
vitamins = data["vitamins"].unique()
```

```python
with open('Dataset_Day2.csv','r') as f:
    r = d.reader(f)
    next(r)
    name = np.array([])
    mfr = np.array([])
    vitamins = np.array([])
    for row in r:
        if row[0] not in name:
            name = np.append(name,row[0])
        if row[1] not in mfr:
            mfr = np.append(mfr,row[1])
        if row[11] not in vitamins:
            vitamins = np.append(vitamins,row[11])
```

```python
# question-2
# Create a new dataframe with all columns for which, ‘sodium’ is greater than 100 AND ‘protein’ is less than 3,
# Name this dataframe:  df_HighSodLowProt

df_HighSodLowProt = pd.DataFrame(data[(data["sodium"]>100) & (data["protein"]<3)])
df_HighSodLowProt
```

```python
# question-3
# From the dataframe in <2.> print the average ‘calories’ by ‘mfr’ and also print the ‘mfr’ with the highest average ‘calories’

# x = df_HighSodLowProt.groupby(['mfr'])['calories'].sum()
# y = df_HighSodLowProt.groupby(['mfr'])['calories'].count()
# z = x / y
z = df_HighSodLowProt.groupby(['mfr'])['calories'].mean()
print('avg ',z)
print(z.index[z.values==z.max()][0],'has maximum average calories')
```
